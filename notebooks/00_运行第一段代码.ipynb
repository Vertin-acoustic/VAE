## 3．变量：给一个值起名字

假设我们正在数哨声，先记下“共有 3 条”。

下面的第一行把整数 `3` 赋给名字 `whistle_count`；第二行把这个名字当前对应的值打印出来。

- `whistle_count` 是我们起的名字，意思是“哨声数量”，不是 Python 特殊命令。
- 下划线 `_` 把两个英文单词连接起来；变量名里不要用空格。
- `=` 在这里表示**赋值**，不是数学推导中的相等关系。可以读作“让这个名字现在对应右边的值”。
- Python 按从上到下的顺序执行这两行。

```python
whistle_count = 3
print(whistle_count)
```

你应该在代码格下方看到 `3`。这个输出是“刚才那次执行的结果”。

## 4．同一个名字，两种写法

先预测下面两行分别会输出什么，再执行。

- 加了英文引号的 `"whistle_count"` 是一段文字，Python 中叫**字符串**。
- 不加引号的 `whistle_count` 表示去查找这个变量当前对应的值。

所以：`print` 可以显示文字，也可以显示变量的值。引号不属于最后显示的文字内容。

```python
print("whistle_count")
print(whistle_count)
```

## 5．内核会记住变量，但不会自动重算

前面的代码格结束了，我们仍可以在新的代码格中使用 `whistle_count`，因为内核记住了它。

下面新增一个变量 `next_count`，表示“当前数量再加 1”。

- `+` 在这段代码里表示数字相加。
- Python 先计算等号右边，再把算出的值赋给左边的名字。
- 这一行执行时完成一次计算，**没有建立自动随源变量更新的公式**。

运行下面的计算格，再运行它下方单独的显示格。

```python
next_count = whistle_count + 1
```

```python
print(next_count)
```

这不是故障。Notebook 允许跳着执行，内核中的值由**实际执行顺序**决定，未必是当前屏幕从上到下的顺序。

## 6．保存文件，与重新开始

- **保存**（通常 `Ctrl+S`）：把修改过的代码、文字和当前输出写入文件。
- **重启内核**（Restart Kernel）：开始新的会话，之前的变量会清空。
- **重新顺序执行**：让新内核按照现在的代码重新建立变量、计算结果。

现在保存文件，重启内核，再从上往下逐格执行所有代码格。最后一个输出应该等于你保留的 `whistle_count` 加 1。

这里只是重新执行现有代码，不要求再次完成文字里的每一步修改。未来遇到“这个格刚才能跑，现在怎么不能跑了”，先检查内核和执行顺序。

**看到保存下来的旧输出，不等于这次会话已经执行过对应代码。**
